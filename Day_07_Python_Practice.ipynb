```python
#FILTER
```

```python
l=[12,34,5,7,8,9,4,6,876,76,54,3,2,0]
```

```python
def even(x):
    if x%2==0:
        return True
    else:
        return False
```

```python
list(filter(even,l))
```

```python
list(filter(lambda x: x%2==0,l))
```

```python
#fetch value from list which is greate than 5
list(filter(lambda x:x>5,l))
```

```python
def big(x):
    if x>5:
        return True
    else:
        return False
```

```python
list(filter(big,l))
```
