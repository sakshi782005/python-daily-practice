```python
#SETS
```

```python
#Create two sets and find their union.
s1=set()
s2=set()
s1={1,2,3,4,6,9,-8}
s2={1,2,5,0,8,6,5,4,3,2}
u=s1|(s2)
print(u)
```

```python
#Find their intersection.
s1=set()
s2=set()
s1={1,2,3,4,6,9,-8}
s2={1,2,5,0,8,6,5,4,3,2}
u=s1&(s2)
print(u)
```

```python
#Find elements present only in the first set.
s1=set()
s2=set()
s1={1,2,3,4,6,9,-8}
s2={1,2,5,0,8,6,5,4,3,2}
u=s2-s1
print(u)
```

```python
#Find symmetric difference.
s1=set()
s2=set()
s1={1,2,3,4,6,9,-8}
s2={1,2,5,0,8,6,5,4,3,2}
u=s1^s2
print(u)
```

```python
#Remove duplicates from a list using a set.
l=[1,2,3,5,7,9,3,5,2,4,3,2,2,9]
print(f"original list:{l}")
s=set(l)
print(f"list convert into set:{s}")
l1=list(s)
print(f"list after removing duplicates{l1}")
```

```python
#DICTONARIES
```

```python
#Create a dictionary containing a student's name, age, branch, and marks.
D={"name":"sakshi",
    "age":21,
    "branch":"entc",
     "marks":89
  }
print(f"dictonary:{D}")
```

```python
#Add a new key called city.
D["city"]="pune"
print(f"dictonary:{D}")
```

```python
#Update the student's marks.
D["marks"]=80
print(f"dictonary:{D}")
```

```python
#Print all keys, values, and key-value pairs.
k=D.keys()
v=D.values()
i=D.items()
print(f"keys:{k}")
print(f"values:{v}")
print(f"key value pairs:{i}")
```

```python
#Find the student with the highest marks.
D={ "sakshi":{
    "age":21,
    "branch":"entc",
     "marks":89
  },
   
   "samir":{
    "age":20,
    "branch":"cs",
     "marks":98
   },

    "prachi":{
    "age":22,
    "branch":"it",
     "marks":95
    }
    }
print(D)

highest_marks=0
for name,info in D.items():
    if info["marks"] > highest_marks:
        highest_marks=info["marks"]
print(f"highest marks:{highest_marks}")
    
```

```python
#Count the frequency of characters in a string using a dictionary.
ch="sakshiBhagat"
d={}

for i in ch:
    if i in d:
        c1=d[i]+1
        d[i]=c1
    else:
        c1=1
        d[i]=c1

print(d)
        
```

```python
#Collections

#Use Counter to count characters in a string.
from collections import Counter
s="SakshiBhagat"
c=Counter(s)
print(c)
```
