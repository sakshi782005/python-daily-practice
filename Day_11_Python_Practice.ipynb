```python
#Create a function to check whether a number is prime.
def prime(n):
    c=0
    for i in range(2,n):
        if n%i==0:
            c=c+1
    if c>2:
        print("not prime")
    else:
        print("prime")
```

```python
n=int(input("enter n"))
prime(n)
```

```python
#Create a function to calculate factorial.
def fact(n):
    factorial=1
    for i in range(n,0,-1):
        factorial=factorial*i
    return factorial
        
```

```python
n=int(input("enter n"))
fact(n)
```

```python
#Create a function to find the largest of three numbers.
def bigN(n1,n2,n3):
    if(n1>n2 and n1>n3):
        print(f"{n1} is largest number among all")
    elif(n2>n1 and n2>n3):
        print(f"{n2} is largest number among all")
    else:
        print(f"{n3} is largest number among all")
```

```python
n1=int(input("enter n1"))
n2=int(input("enter n2"))
n3=int(input("enter n3"))
bigN(n1,n2,n3)
```

```python
#MAP
```

```python
l=[1,2,3,7,6,4,9,8,3]
```

```python
#Use map() to square every number in a list.
sqr=list(map(lambda n:n**2,l))
print(sqr)
```

```python
#Use map() to add 10 to every number.
add=list(map(lambda n:n+10,l))
print(add)
```

```python
#Convert a list of strings into integers using map().
l1=['145','102','820','88','52','98']
convert=list(map(lambda n:int(n),l1))
print(convert)
```

```python
#FILTER
```

```python
#Use filter() to get even numbers.
l=[1,2,3,7,6,4,9,8,3]
even=list(filter(lambda n:n%2==0,l)) 
print(even)
```

```python
#Use filter() to get numbers greater than 50.
l=[1,5,8,6,98,75,20,38,34,80,50,72,698,152,753,148,24,86]
check=list(filter(lambda n:n>50,l))
print(check)
```

```python
#Use filter() to get words having more than 4 characters.
l2=['asdfg','asdrcvg','as','erc','tg','qwert']
words=list(filter(lambda ch:len(ch)>4,l2))
print(words)
```

```python
#Given a list of numbers, use map() + filter() to:
#first select even numbers
#then square them.
l=[1,5,8,6,98,75,20,38,34,80,50,72,698,152,753,148,24,86,852,7896,14,2,2034]
m=list(filter(lambda n:n%2==0,l))
sqr=list(map(lambda n:n**2,m))
print(f"even numbers list:{m}")
print(f"square of even number list:{sqr}")
```

```python
#Filter words starting with 'a'
words = ['apple', 'banana', 'ant', 'cat', 'avocado']
fil=list(filter(lambda ch:ch.startswith('a'),words))
print(fil)
fil=list(filter(lambda ch:ch.endswith('a'),words))
print(fil)
```

```python
#Filter words with even length
words = ['book', 'cat', 'tree', 'apple', 'hi']
even_len=list(filter(lambda ch:len(ch)%2==0,words))
print(even_len)
```

```python
#Get the length of every word
words = ['apple', 'kiwi', 'banana']
len=list(map(lambda ch:len(ch),words))
print(len)
```

```python
#Keep numbers divisible by 3 and 5
numbers = [15, 10, 30, 9, 45, 22, 60]
div=list(filter(lambda n:n%3==0 and n%5==0,numbers))
print(div)
```

```python
#Count vowels in a word
word="sakshibhagat"
count=0
for ch in word:
    if ch in['a','e','i','o','u']:
        count=count+1
print(count)
```
