```python
def isprime(n):
    for i in range(2,n):
        if(n%i==0):
            return False
        else:
            return True
```

```python
def absolutevaluse(n):
    if n>0:
        return n
    else:
        return n*-1
```

```python
def sqrt(n):
    return n**0.5
```

```python
def LEN(data):
    c=0
    for i in data:
        c+=1
    return c
```
