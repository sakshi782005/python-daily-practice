```python
#print number using while loop 1)take input from user 
#1input:starting number 2input:ending number
start=int(input("enter starting number"))
end=int(input("enter ending number"))
while(start<=end):
    print(start)
    start=start+1
    
```

```python
#print number but if given starting number is bigger than ending then print reverse and if not then print normal order
start=int(input("enter starting point"))
end=int(input("enter ending point"))
if(start<end):
    while(start<=end):
        print(start)
        start=start+1
elif(start>end):
    while(start>=end):
        print(start)
        start=start-1
else:
    print(start)
```

```python
#Write all the prime numbers between 1 and 20.
i=1
while(i<=20):
    count=0
    j=1
    while(j<=i):
        if(i%j==0):
            count=count+1
        j=j+1
    if(count==2):
        print(f"prime number: {i}")
    i=i+1
    
```

```python
#Write all the composite numbers between 21 and 50.
i=21
while(i<=50):
    count=0
    j=1
    while(j<=i):
        if(i%j==0):
            count=count+1
        j=j+1
    if(count>2):
        print(f"composite number :{i}")
    i=i+1;
```

```python
#Which of the prime numbers are even numbers ?
i=1
while(i<100):
    count=0
    j=1
    while(j<=i):
        if(i%j==0):
            count=count+1
        j=j+1
    if(count==2 and i%2==0):
        print(f"{i} is a prime and even number")
    i=i+1
```
