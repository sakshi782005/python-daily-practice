```python
#Take a string and print its first and last character.
s="sakshi"
l=len(s)
for i in range(l):
    if(i==0 or i==l-1):
        print(s[i])
```

```python
#Reverse a string without using a built-in reverse function. 
s="sakshi"
l=len(s)
for i in range(l-1,-1,-1):
    print(s[i])
```

```python
#Count the number of vowels in a string.
s=input("enter string")
l=len(s)
count=0
for n in range(l):
    if s[n]=='a' or s[n]=='e' or s[n]=='i' or s[n]=='o' or s[n]=='u' :
        count+=1
print(count)
        
```

```python
#Count how many times "a" occurs in a string.
s=input("enter string")
l=len(s)
count=0
for i in range(l):
    if(s[i]=='a'):
        count+=1
print(count)
```

```python
#Check whether a string is a palindrome.
s=input("enter string")
l=len(s)
for i in range(l//2):
    if s[i]==s[l-1-i]:
        print("palindrome")
    else:
        print("not palindrome")
    
```

```python
#Take a sentence and count the number of words.
s=input("enter sentence")
s1=s.split()
print(s1)
l=len(s1)
print(f"number of words in sentence is :{l}")
```

```python
#Take a string and print: total characters vowels consonants digits
s=input("enter string")
l=len(s)
count_v=0
count_c=0

print(f"total character:{l}")

for i in range(l):
    if(s[i]=='a'or s[i]=='e'or s[i]=='i'or s[i]=='o'or s[i]=='u'):
        count_v= count_v+1
    
    else:
        count_c=count_c+1
        
print(f"count of vowel: {count_v}")
print(f"count of consonants: {count_c}")
```

```python
#Count each vowel separately
s=input("enter string:")
l=len(s)
c_a=0
c_e=0
c_i=0
c_o=0
c_u=0
for n in range(l):
    if s[n]=='a':
        c_a+=1
    elif s[n]=='e':
        c_e+=1
    elif s[n]=='i':
        c_i+=1
    elif s[n]=='o':
        c_o+=1
    elif s[n]=='u':
        c_u+=1
print(f"count \n a:{c_a} \n e:{c_e} \n i:{c_i} \n o:{c_o} \n u:{c_u}")
```
