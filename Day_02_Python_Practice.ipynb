```python
#acending order
list=[2,5,9,88,7,3,66,87,85,55,472,38,9999,1,0,7]
n=len(list)
for i in range(n):
    for j in range(n-1):
        if list[j]>list[j+1]:
            list[j],list[j+1]=list[j+1],list[j]
print(list)
```

```python
#decending order
list=[2,5,9,88,7,3,66,87,85,55,472,38,9999,1,0,7]
n=len(list)
for i in range(n):
    for j in range(n-1):
        if list[j+1]>list[j]:
            list[j+1],list[j]=list[j],list[j+1]
print(list)
```

```python
#take input from user and then sort them according to their preferece like acending or decending
list=[]
n=int(input("enter how many number you want to add in list"))
for i in range(n):
    n1=int(input(f"enter {i+1}_number"))
    list.append(n1)  
print(list)
method=int(input("if you want ascending sort give input:1 , descending:2, both:3"))
if method==1:
    for i in range(n):
        for j in range(n-1):
            if list[j]>list[j+1]:
                list[j],list[j+1]=list[j+1],list[j]
    print(f"ascending order: {list}")

elif method==2:
    for i in range(n):
        for j in range(n-1):
            if list[j+1]>list[j]:
                list[j+1],list[j]=list[j],list[j+1]
    print(f"descending order: {list}")

else:
    for i in range(n):
        for j in range(n-1):
            if list[j]>list[j+1]:
                list[j],list[j+1]=list[j+1],list[j]
    print(f"ascending order: {list}")

    for i in range(n):
        for j in range(n-1):
            if list[j+1]>list[j]:
                list[j+1],list[j]=list[j],list[j+1]
    print(f"descending order: {list}")
```

```python
start=int(input("enter the satrting number"))
end=int(input("enter the ending number"))
even=[]
odd=[]
for i in range(start,end+1):
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
print(f"even numbers between {start} to {end}: {even}")
print(f"odd numbers between {start} to {end}: {odd}")
```
