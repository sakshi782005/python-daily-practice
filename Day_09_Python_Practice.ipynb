```python
#Separate positive, negative and zero
numbers = [5, -2, 0, 8, -7, 0, 3, -1]
positive=[]
negative=[]
zero=[]
l=len(numbers)
for n in numbers:
    if n>0:
        positive.append(n)
    elif n<0:
        negative.append(n)
    else:
        zero.append(n)
print(f"positive number's in list:{positive}")
print(f"negative number's in list:{negative}")
print(f"zero's in list:{zero}")
```
