```python
#Write all the factors of the given numbers 
n1=int(input("enter number"))
n2=int(input("enter number"))
if(n1%n2==0):
    print(f"{n2} is factor of {n1}")
else:
    print(f"{n2} is not a factor of {n1}")
```

```python
#Write all the factors of the given numbers and list their common factors.
n1=int(input("enter number"))
n2=int(input("enter number"))
i=1
f1=[]
f2=[]
while(i<=50):
    if n1%i==0:
        f1.append(i)
    if n2%i==0:
        f2.append(i)
    i=i+1
print(f1)
print(f2)
```
